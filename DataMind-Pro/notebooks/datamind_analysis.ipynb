## 自动分析摘要\n- 记录数：7\n- 字段数：6\n- 收入字段：销售额\n- 收入规模：¥97,400\n- 利润率：56.6%\n

```python
import csv, statistics, pathlib

path = pathlib.Path('../data/sample-business-finance.csv')
rows = list(csv.DictReader(path.open(encoding='utf-8-sig')))

def num(value):
    try:
        return float(str(value).replace(',', ''))
    except Exception:
        return None

revenue_col = '销售额'
cost_col = '成本'
revenue = [num(r[revenue_col]) for r in rows if revenue_col and num(r[revenue_col]) is not None]
cost = [num(r[cost_col]) for r in rows if cost_col and num(r[cost_col]) is not None]
profit = sum(revenue) - sum(cost)
margin = profit / sum(revenue) if revenue else None

print('records:', len(rows))
print('columns:', list(rows[0].keys()))
print('revenue:', round(sum(revenue), 2))
print('profit:', round(profit, 2))
print('margin:', round(margin * 100, 2), '%')
```
